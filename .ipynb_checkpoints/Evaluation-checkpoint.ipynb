```python
import pandas as pd
import json
from collections import Counter
from sklearn.metrics import adjusted_rand_score, normalized_mutual_info_score, v_measure_score

# Load sample with ground truth
# sample_df = pd.read_csv("sample.csv")

# Micro-averaged accuracy helper
def microavg_within_cluster_accuracy(df, true_col="manual_group_id", pred_col="predicted_cluster"):
    total_correct = 0
    total_items = len(df)

    for _, cluster_df in df.groupby(pred_col):
        if len(cluster_df) == 0:
            continue
        true_labels = cluster_df[true_col]
        most_common_label_count = Counter(true_labels).most_common(1)[0][1]
        total_correct += most_common_label_count

    return total_correct / total_items if total_items > 0 else 0

# Evaluation function
def evaluate_clustering(json_path, sample_df, true_col="manual_group_id"):
    with open(json_path, "r") as f:
        clusters_by_group = json.load(f)

    records = []
    for group_key, cluster_dict in clusters_by_group.items():
        for cluster_id_str, items in cluster_dict.items():
            for item in items:
                records.append({
                    "mirakl_product_id": item["product_id"],
                    "predicted_cluster": int(cluster_id_str)
                })

    pred_df = pd.DataFrame(records)

    # Merge with ground truth
    merged = pd.merge(pred_df, sample_df[["mirakl_product_id", true_col]], on="mirakl_product_id", how="inner")
    merged = merged.dropna(subset=[true_col])

    y_true = merged[true_col].astype(int).values
    y_pred = merged["predicted_cluster"].values

    # Metrics
    ari = adjusted_rand_score(y_true, y_pred)
    nmi = normalized_mutual_info_score(y_true, y_pred)
    vms = v_measure_score(y_true, y_pred)
    micro_acc = microavg_within_cluster_accuracy(merged, true_col=true_col, pred_col="predicted_cluster")

    print(f"📂 {json_path}")
    print(f"   Adjusted Rand Index (ARI): {ari:.3f}")
    print(f"   Normalized Mutual Info (NMI): {nmi:.3f}")
    print(f"   V-measure: {vms:.3f}")
    print(f"   Micro-averaged Accuracy: {micro_acc:.3f}")
    print("-" * 60)
```

### Image Feature extraction

```python
import torch
import requests
import numpy as np
import pandas as pd
from PIL import Image
from io import BytesIO
from tqdm import tqdm
from sklearn.preprocessing import normalize
from sklearn.metrics.pairwise import cosine_similarity
import open_clip
from torchvision import transforms

# Load sample
sample_df = pd.read_csv("sample.csv")

# Load CLIP model and preprocessing
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model, _, preprocess = open_clip.create_model_and_transforms('ViT-B-32', pretrained='openai')
model = model.to(device).eval()

# Modified transform to force grayscale
grayscale_preprocess = transforms.Compose([
    transforms.Grayscale(num_output_channels=3),
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.48145466, 0.4578275, 0.40821073],
        std=[0.26862954, 0.26130258, 0.27577711]
    )
])

# Image columns to try
fallback_cols = [
    'mainImage.original_url',
    'secondary_image_1.original_url',
    'secondary_image_2.original_url',
    'secondary_image_3.original_url'
]

# Extract grayscale CLIP embeddings
image_embeddings = []
image_ids = []
broken_image_rows = []

for idx, row in tqdm(sample_df.iterrows(), total=len(sample_df)):
    success = False
    for col in fallback_cols:
        url = row.get(col)
        if not isinstance(url, str) or not url.startswith("http"):
            continue

        try:
            response = requests.get(url, timeout=10)
            if response.status_code != 200 or b"<html" in response.content[:100].lower():
                raise ValueError("Not a valid image response")

            img = Image.open(BytesIO(response.content)).convert("RGB")
            img_gray = grayscale_preprocess(img).unsqueeze(0).to(device)

            with torch.no_grad():
                features = model.encode_image(img_gray).cpu().squeeze().numpy()

            image_embeddings.append(features)
            image_ids.append(row["mirakl_product_id"])
            success = True
            break

        except Exception as e:
            continue

    if not success:
        broken_image_rows.append({
            "row_idx": idx,
            "product_id": row.get("mirakl_product_id"),
            "last_url_attempted": url
        })

# Save broken image info
pd.DataFrame(broken_image_rows).to_csv("broken_images.csv", index=False)

# Save extracted embeddings
np.save("grayscale_clip_embeddings_sample.npy", np.array(image_embeddings))
pd.DataFrame({"mirakl_product_id": image_ids}).to_csv("grayscale_clip_ids_sample.csv", index=False)

print("✅ Grayscale CLIP embeddings extracted and saved.")
```

## An attempt to find variants using both groupby operations and textual embeddings as well as image embeddings (only brand and category) (call this method *)

```python
import numpy as np
import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.preprocessing import normalize
import hdbscan
from collections import defaultdict
import json

# === Step 1: Load CSV and grayscale image embeddings ===
sample_df = pd.read_csv("sample.csv")
image_vecs_all = np.load("grayscale_clip_embeddings_sample.npy")
image_ids = pd.read_csv("grayscale_clip_ids_sample.csv")["mirakl_product_id"].tolist()

# === Step 2: Align sample_df with loaded image IDs ===
df_clip = pd.DataFrame({
    "mirakl_product_id": image_ids,
    "clip_embedding": list(image_vecs_all)
})
sample_df = pd.merge(sample_df, df_clip, on="mirakl_product_id", how="inner")

# === Step 3: Define grouping keys (brand + category) ===
group_keys = ['brand', 'category']
group_sizes = sample_df.groupby(group_keys).size().reset_index(name='count')
duplicates = group_sizes[group_sizes["count"] > 1]
identical_groups = sample_df.merge(duplicates[group_keys], on=group_keys, how="inner")
identical_groups["product_group_id"] = identical_groups.groupby(group_keys).ngroup()

# === Step 4: Load text model ===
text_model = SentenceTransformer("all-MiniLM-L6-v2")

# === Step 5: Cluster per group ===
results_all = {}

for group_id, group_df in identical_groups.groupby("product_group_id"):
    # Extract text fields
    names = group_df["name [en_GB]"].fillna("")
    brands = group_df["brand"].fillna("")
    colors = group_df["color"].fillna("")
    descriptions = group_df["shortDescription [en_GB]"].fillna("")
    texts = (names + " " + brands + " " + colors + " " + descriptions).tolist()
    text_vecs = text_model.encode(texts, normalize_embeddings=True)

    # Extract grayscale image embeddings
    img_vecs = np.vstack(group_df["clip_embedding"].values)
    img_vecs = normalize(img_vecs)

    # Combine
    alpha = 0.6
    combined_vecs = np.hstack([
        alpha * img_vecs,
        (1 - alpha) * text_vecs
    ])
    
    combined_vecs = normalize(combined_vecs)


    # Cluster
    clusterer = hdbscan.HDBSCAN(min_cluster_size=2, metric='euclidean')
    labels = clusterer.fit_predict(combined_vecs)

    # Format output
    group_df = group_df.copy()
    group_df["cluster_id"] = labels
    grouped_json = defaultdict(list)

    for _, row in group_df.iterrows():
        grouped_json[int(row["cluster_id"])].append({
            "product_id": row["mirakl_product_id"],
            "name": row["name [en_GB]"],
            "image": row["mainImage.original_url"]
        })

    results_all[f"group_{group_id}"] = grouped_json

# === Step 6: Save to JSON ===
with open("clustered_duplicate_groups_sample.json", "w") as f:
    json.dump(results_all, f, indent=2)

print("✅ Done! Grayscale CLIP + Text hybrid clustering complete.")
```

```python
evaluate_clustering("clustered_duplicate_groups_sample.json", sample_df)
```

## An attempt to find variants using both groupby operations and textual embeddings as well as image embeddings (only brand and category and name en_gb)(call this method **)

```python
import numpy as np
import pandas as pd
import json
from sentence_transformers import SentenceTransformer
from sklearn.preprocessing import normalize
import hdbscan
from collections import defaultdict

# === Step 1: Load flat CSV and grayscale image embeddings ===
sample_df = pd.read_csv("sample.csv")
image_vecs_all = np.load("grayscale_clip_embeddings_sample.npy")
sample_df = sample_df.reset_index(drop=True).iloc[:len(image_vecs_all)]  # align rows with embeddings

# === Step 2: Define grouping keys (strict) ===
group_keys = ['brand', 'category', 'name [en_GB]']

# === Step 3: Identify duplicate metadata groups ===
group_sizes = sample_df.groupby(group_keys).size().reset_index(name='count')
duplicates = group_sizes[group_sizes["count"] > 1]
identical_groups = sample_df.merge(duplicates[group_keys], on=group_keys, how="inner")
identical_groups["product_group_id"] = identical_groups.groupby(group_keys).ngroup()

# === Step 4: Load sentence transformer ===
text_model = SentenceTransformer("all-MiniLM-L6-v2")

# === Step 5: Clustering per group ===
results_all = {}

for group_id, group_df in identical_groups.groupby("product_group_id"):
    indices = group_df.index.tolist()

    # Text input
    names = group_df["name [en_GB]"].fillna("")
    brands = group_df["brand"].fillna("")
    colors = group_df["color"].fillna("")
    descriptions = group_df["shortDescription [en_GB]"].fillna("")
    texts = (names + " " + brands + " " + colors + " " + descriptions).tolist()
    text_vecs = text_model.encode(texts, normalize_embeddings=True)

    # Image input
    img_vecs = normalize(image_vecs_all[indices])

    # Combine both
    alpha = 0.6
    combined_vecs = np.hstack([alpha * img_vecs, (1 - alpha) * text_vecs])
    combined_vecs = normalize(combined_vecs)

    # Clustering
    clusterer = hdbscan.HDBSCAN(min_cluster_size=2, metric='euclidean')
    labels = clusterer.fit_predict(combined_vecs)

    group_df = group_df.copy()
    group_df["cluster_id"] = labels

    # Store results as JSON
    grouped_json = defaultdict(list)
    for _, row in group_df.iterrows():
        grouped_json[int(row["cluster_id"])].append({
            "product_id": row.get("mirakl_product_id"),
            "name": row.get("name [en_GB]", ""),
            "image": row.get("mainImage.original_url", "")
        })

    results_all[f"group_{group_id}"] = grouped_json

# === Step 6: Save all cluster groups ===
with open("clustered_duplicate_groups_2_sample.json", "w") as f:
    json.dump(results_all, f, indent=2)

print("✅ Done! Combined image + text clustering finished on sample.")
```

## Comparing this to using no images - just text metadata embeddings (finding variants - do not include colour as a group key)

```python
import numpy as np
import pandas as pd
import json
from sentence_transformers import SentenceTransformer
from sklearn.preprocessing import normalize
import hdbscan
from collections import defaultdict
from IPython.display import display, HTML

# === Step 1: Load flat CSV ===
sample_df = pd.read_csv("sample.csv")

# === Step 2: Group using metadata keys ===
group_keys = ['brand', 'category', 'name [en_GB]']
group_sizes = sample_df.groupby(group_keys).size().reset_index(name='count')
duplicates = group_sizes[group_sizes["count"] > 1]
identical_groups = sample_df.merge(duplicates[group_keys], on=group_keys, how="inner")
identical_groups["product_group_id"] = identical_groups.groupby(group_keys).ngroup()

# === Step 3: Load sentence-transformer ===
text_model = SentenceTransformer("all-MiniLM-L6-v2")

# === Step 4: Cluster per metadata group ===
results_all = {}

for group_id, group_df in identical_groups.groupby("product_group_id"):
    # Combine text features
    names = group_df["name [en_GB]"].fillna("")
    brands = group_df["brand"].fillna("")
    colors = group_df["color"].fillna("")
    descriptions = group_df["shortDescription [en_GB]"].fillna("")
    texts = (names + " " + brands + " " + colors + " " + descriptions).tolist()

    # Encode and normalize
    text_vecs = text_model.encode(texts, normalize_embeddings=True)

    # Run HDBSCAN
    clusterer = hdbscan.HDBSCAN(min_cluster_size=2, metric='euclidean')
    labels = clusterer.fit_predict(text_vecs)

    group_df = group_df.copy()
    group_df["cluster_id"] = labels

    # Store JSON structure
    grouped_json = defaultdict(list)
    for _, row in group_df.iterrows():
        grouped_json[int(row["cluster_id"])].append({
            "product_id": row.get("mirakl_product_id"),
            "name": row.get("name [en_GB]"),
            "image": row.get("mainImage.original_url", "")
        })

    results_all[f"group_{group_id}"] = grouped_json

# === Step 5: Save results ===
with open("text_only_clustered_groups_2_sample.json", "w") as f:
    json.dump(results_all, f, indent=2)

print("✅ Text-only clustering complete!")

# === Step 6: Visualize clusters ===
with open("text_only_clustered_groups_2_sample.json", "r") as f:
    clusters_by_group = json.load(f)

def show_group_cluster(group_id, cluster_id=None, max_items=6):
    group_key = f"group_{group_id}"
    if group_key not in clusters_by_group:
        print(f"Group {group_id} not found.")
        return

    clusters = clusters_by_group[group_key]
    cluster_ids = [cluster_id] if cluster_id is not None else list(clusters.keys())

    for cid in cluster_ids:
        if str(cid) not in clusters:
            print(f"  Cluster {cid} not found in {group_key}")
            continue
        items = clusters[str(cid)][:max_items]
        html = f"<h3 style='margin-top:20px'>Group {group_id} – Cluster {cid} ({len(clusters[str(cid)])} items)</h3><div style='display:flex;flex-wrap:wrap'>"
        for item in items:
            name = item.get('name', '')
            img = item.get('image', '')
            html += f"""
                <div style='margin:10px;text-align:center'>
                    <img src='{img}' width='120' style='border-radius:8px'><br>
                    <div style='max-width:120px;font-size:12px'>{name}</div>
                </div>
            """
        html += "</div>"
        display(HTML(html))

# === Example usage ===
show_group_cluster(0)
show_group_cluster(1, cluster_id=0)
```

## Comparing this to using no images - just text metadata embeddings (finding variants - do not include colour as a group key) - but this time excluding  nameen_gb as a group key

```python
import numpy as np
import pandas as pd
import json
from sentence_transformers import SentenceTransformer
from sklearn.preprocessing import normalize
import hdbscan
from collections import defaultdict
from IPython.display import display, HTML

# === Step 1: Load CSV ===
sample_df = pd.read_csv("sample.csv")

# === Step 2: Group by metadata keys (excluding image info) ===
group_keys = ['brand', 'category']
group_sizes = sample_df.groupby(group_keys).size().reset_index(name='count')
duplicates = group_sizes[group_sizes["count"] > 1]
identical_groups = sample_df.merge(duplicates[group_keys], on=group_keys, how="inner")
identical_groups["product_group_id"] = identical_groups.groupby(group_keys).ngroup()

# === Step 3: Load sentence transformer ===
text_model = SentenceTransformer("all-MiniLM-L6-v2")

# === Step 4: Cluster text-only features per group ===
results_all = {}

for group_id, group_df in identical_groups.groupby("product_group_id"):
    # Extract text fields
    names = group_df["name [en_GB]"].fillna("")
    brands = group_df["brand"].fillna("")
    colors = group_df["color"].fillna("")
    descriptions = group_df["shortDescription [en_GB]"].fillna("")
    texts = (names + " " + brands + " " + colors + " " + descriptions).tolist()

    # Encode and cluster
    text_vecs = text_model.encode(texts, normalize_embeddings=True)
    clusterer = hdbscan.HDBSCAN(min_cluster_size=2, metric='euclidean')
    labels = clusterer.fit_predict(text_vecs)
    group_df = group_df.copy()
    group_df["cluster_id"] = labels

    # Save clusters
    grouped_json = defaultdict(list)
    for _, row in group_df.iterrows():
        grouped_json[int(row["cluster_id"])].append({
            "product_id": row.get("mirakl_product_id"),
            "name": row.get("name [en_GB]", ""),
            "image": row.get("mainImage.original_url", "")
        })

    results_all[f"group_{group_id}"] = grouped_json

# === Step 5: Save result to JSON ===
with open("text_only_clustered_groups_3_sample.json", "w") as f:
    json.dump(results_all, f, indent=2)

print("✅ Text-only clustering complete!")

# === Step 6: Visualize results ===
with open("text_only_clustered_groups_3_sample.json", "r") as f:
    clusters_by_group = json.load(f)

def show_group_cluster(group_id, cluster_id=None, max_items=6):
    group_key = f"group_{group_id}"
    if group_key not in clusters_by_group:
        print(f"Group {group_id} not found.")
        return

    clusters = clusters_by_group[group_key]
    cluster_ids = [cluster_id] if cluster_id is not None else list(clusters.keys())

    for cid in cluster_ids:
        if str(cid) not in clusters:
            print(f"  Cluster {cid} not found in {group_key}")
            continue
        items = clusters[str(cid)][:max_items]
        html = f"<h3 style='margin-top:20px'>Group {group_id} – Cluster {cid} ({len(clusters[str(cid)])} items)</h3><div style='display:flex;flex-wrap:wrap'>"
        for item in items:
            name = item.get('name', '')
            img = item.get('image', '')
            html += f"""
                <div style='margin:10px;text-align:center'>
                    <img src='{img}' width='120' style='border-radius:8px'><br>
                    <div style='max-width:120px;font-size:12px'>{name}</div>
                </div>
            """
        html += "</div>"
        display(HTML(html))

# === Example usage ===
show_group_cluster(0)
show_group_cluster(1, cluster_id=0)
```
